```python
%%capture
# environment set up
# necessary packages

from concurrent.futures import ThreadPoolExecutor

def setup_node(node):
    """
    Function to execute the setup commands on a given node.
    """
    try:
        # List of commands to be executed sequentially on each node
        commands = [
            'sudo apt install software-properties-common',
            'sudo add-apt-repository ppa:named-data/ppa',
            'sudo apt update',
            'sudo apt install nfd',
            'sudo apt -y install python3-pip libndn-cxx-dev nfd ndnping ndnpeek ndn-dissect ndnchunks ndnsec jq',
            'sudo add-apt-repository ppa:deadsnakes/ppa',
            'sudo apt update',
            'sudo apt -y install python3.8 python3.8-distutils',
            'wget https://bootstrap.pypa.io/get-pip.py',
            'sudo python3.8 get-pip.py',
            'git clone https://github.com/tntech-ngin/ndn-hydra.git',
            'cd ndn-hydra/ && python3.8 -m pip install -e .',
            'python3.8 -m pip install python-ndn ndn-storage ndn-svs numpy',
            'sudo apt install net-tools',
            'sudo apt update && sudo apt upgrade -y',
            'sudo cp /etc/ndn/nfd.conf.sample /etc/ndn/nfd.conf',
            'cd ~/ndn-hydra',
            'git checkout feature/ndn-hydra-dpdk',
            'git pull origin feature/ndn-hydra-dpdk',
            'python3.8 -m pip install -e .'
            'sudo chown -R ubuntu:ubuntu /home/ubuntu/.ndn/repo'
        ]
        
        # Execute commands in sequence for the node
        for cmd in commands:
            stdout, stderr = node.execute(cmd)
        
        return f"Node {node.get_name()} setup completed successfully."
    except Exception as e:
        return f"Node {node.get_name()} failed with exception: {e}"

# Main execution
try:
    slice_nodes = slice.get_nodes()
    
    # Using ThreadPoolExecutor for parallel execution
    with ThreadPoolExecutor(max_workers=len(slice_nodes)) as executor:
        # Map the setup_node function to all nodes
        executor.map(setup_node, slice_nodes)
except Exception as e:
    print(f"Exception during parallel execution: {e}")
```

```python
%%capture
# Bring the interfaces up
from concurrent.futures import ThreadPoolExecutor

def bring_interfaces_up(node):
    """
    Bring up all interfaces on a node, excluding 'uplink4'.
    """
    try:
        interfaces = node.get_interfaces()
        for interface in interfaces:
            if 'uplink4' not in interface.get_name():
                print(f"Bringing up interface: {interface.get_name()} on node: {node.get_name()}")
                stdout, stderr = node.execute(f'sudo ip link set dev {interface.get_device_name()} up')
                stdout, stderr = node.execute('sleep 2; ip a')
        return f"Interfaces brought up for node: {node.get_name()}"
    except Exception as e:
        return f"Error bringing up interfaces for node {node.get_name()}: {e}"

def setup_nfd(node, role, index):
    """
    Stop, start NFD, and generate keys for a node.
    """
    try:
        print(f"Setting up NFD on {node.get_name()} ({role})...")
        stdout, stderr = node.execute('nfd-stop')
        stdout, stderr = node.execute('nfd-start')
        stdout, stderr = node.execute('sleep 2; nfd-status')
        
        if role == "client":
            key_command = f'ndnsec key-gen /client{index + 1} | ndnsec cert-install -'
        else:
            key_command = f'ndnsec key-gen /hydra/node{index + 1} | ndnsec cert-install -'
        
        stdout, stderr = node.execute(key_command)
        return f"NFD setup completed for {node.get_name()} ({role})"
    except Exception as e:
        return f"Error setting up NFD for node {node.get_name()}: {e}"

# Main execution
try:
    # Bring interfaces up for all nodes
    print("Bringing interfaces up...")
    all_nodes = slice.get_nodes()
    with ThreadPoolExecutor(max_workers=len(all_nodes)) as executor:
        executor.map(bring_interfaces_up, all_nodes)

    # Set up NFD for clients
    print("Setting up NFD for clients...")
    clients = [slice.get_node(client1_name), slice.get_node(client2_name), slice.get_node(client3_name)]
    with ThreadPoolExecutor(max_workers=len(clients)) as executor:
        executor.map(setup_nfd, clients, ["client"] * len(clients), range(len(clients)))

    # Set up NFD for Hydra nodes
    print("Setting up NFD for Hydra nodes...")
    hydra_nodes = [
        slice.get_node(node_a_name),
        slice.get_node(node_b_name),
        slice.get_node(node_c_name),
        slice.get_node(node_d_name),
        slice.get_node(node_e_name),
        slice.get_node(node_f_name),
        slice.get_node(node_g_name),
    ]
    with ThreadPoolExecutor(max_workers=len(hydra_nodes)) as executor:
        executor.map(setup_nfd, hydra_nodes, ["hydra"] * len(hydra_nodes), range(len(hydra_nodes)))

except Exception as e:
    print(f"Exception during execution: {e}")
```

```python
%%capture
from concurrent.futures import ThreadPoolExecutor

# NDN-DPDK git repository
NDNDPDK_GIT = ndndpdk_common.DEFAULT_GIT_REPO

def setup_initial_node(node):
    """
    Perform the initial setup for a node (install NDN-DPDK, reboot, etc.).
    """
    try:
        node.execute(f'''
            echo 'set enable-bracketed-paste off' | sudo tee -a /etc/inputrc
            sudo hostnamectl set-hostname {node.get_name()}
            {ndndpdk_common.apt_install_cmd(extra_pkgs=[])}
            sudo loginctl enable-linger {node.get_username()}
            sudo systemctl reboot
        ''')
        return f"Initial setup completed on {node.get_name()}."
    except Exception as e:
        return f"Initial setup failed on {node.get_name()}: {e}"

def setup_ndndpdk_node(node):
    """
    Build and install NDN-DPDK, and configure CPU isolation for a node.
    """
    try:
        node.execute(ndndpdk_common.dl_build_cmd(repo=NDNDPDK_GIT))
        node.execute(f"""
            {ndndpdk_common.cpuset_cmd(node, instances={
                '127.0.0.1:3030': 10,
                '127.0.0.1:3031': 8,
            })}
            sudo systemctl reboot
        """)
        return f"NDN-DPDK setup completed on {node.get_name()}."
    except Exception as e:
        return f"NDN-DPDK setup failed on {node.get_name()}: {e}"

# Main execution
try:
    # Retrieve all nodes
    node_names = [node_a_name, node_b_name, node_c_name, node_d_name, node_e_name, node_f_name, node_g_name]
    nodes = [slice.get_node(name) for name in node_names]
    
    # Phase 1: Initial setup and reboot
    with ThreadPoolExecutor(max_workers=len(nodes)) as executor:
        executor.map(setup_initial_node, nodes)

    # Wait for all nodes to become SSH-accessible again
    slice.wait_ssh(progress=True)
    
    # Phase 2: Build and install NDN-DPDK, and configure CPU isolation
    with ThreadPoolExecutor(max_workers=len(nodes)) as executor:
        executor.map(setup_ndndpdk_node, nodes)
    
    # Wait for all nodes to reboot and become SSH-accessible again
    slice.wait_ssh(progress=True)

except Exception as e:
    print(f"Exception during parallel execution: {e}")
```

```python
from concurrent.futures import ThreadPoolExecutor
import json
import shlex

CTRL_FW = 'ndndpdk-ctrl --gqlserver http://127.0.0.1:3030/'
CTRL_TG = 'ndndpdk-ctrl --gqlserver http://127.0.0.1:3032/'

FW_ACTIVATE = {
    'eal': {
        'memPerNuma': {'0': 20*1024},
        'coresPerNuma': {'0': 10}
    },
    'mempool': {
        'DIRECT': {'capacity': 2**20-1, 'dataroom': 9200},
        'INDIRECT': {'capacity': 2**21-1},
    },
    'lcoreAlloc': {
        'RX': {'0': 2},
        'TX': {'0': 2},
        'CRYPTO': {'0': 2},
        'FWD': {'0': 3}
    }
}

MEMIF_SOCKET_TG, MEMIF_ID_TG = '/run/ndn/tg.sock', 1
FW_MEMIF_TG = {
    'scheme': 'memif',
    'socketName': MEMIF_SOCKET_TG,
    'id': MEMIF_ID_TG,
    'role': 'server',
    'dataroom': 9000
}

TG_ACTIVATE = {
    'mempool': {
        'DIRECT': {'capacity': 65535, 'dataroom': 9128},
        'INDIRECT': {'capacity': 65535},
    },
}


def setup_forwarder(node):
    """
    Set up the forwarder and NDNc on a client node.
    """
    try:
        # activate NDN-DPDK forwarder, and trafficgen
        node.execute(f'''
            sudo {CTRL_FW} systemd stop || true
            sudo {CTRL_TG} systemd stop || true
        
            {ndndpdk_common.hugepages_cmd(size=25)}
            sudo {CTRL_FW} systemd start
            echo {shlex.quote(json.dumps(FW_ACTIVATE))} | {CTRL_FW} activate-forwarder

            FW_MEMIF_FACE=$(echo {shlex.quote(json.dumps(FW_MEMIF_TG))} | {CTRL_FW} create-face)
            echo $FW_MEMIF_FACE
        
            sudo {CTRL_TG} systemd start
            echo {shlex.quote(json.dumps(TG_ACTIVATE))} | {CTRL_TG} activate-trafficgen
        ''')
        
        # Create ethernet ports
        for interface in node.get_interfaces():
            if '_dpdk' in interface.get_name():
                node.execute(f'''
                    ndndpdk-ctrl create-eth-port --pci $(basename $(realpath /sys/class/net/{shlex.quote(interface.get_physical_os_interface_name())}/device)) --mtu 9000 --rx-flow 16
                ''')
        
        return f"Forwarder setup completed on {node.get_name()}."
    except Exception as e:
        return f"Error on {node.get_name()}: {e}"

# Main execution
try:
    # Retrieve client nodes
    client_names = [client1_name, client2_name, client3_name]
    clients = [slice.get_node(name) for name in client_names]
    
    # Perform setup on all clients in parallel
    with ThreadPoolExecutor(max_workers=len(clients)) as executor:
        results = list(executor.map(setup_forwarder, clients))

except Exception as e:
    print(f"Exception during parallel execution: {e}")
```
